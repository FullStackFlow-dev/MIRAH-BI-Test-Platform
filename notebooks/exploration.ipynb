# Restaurant BI Project

Exploration des données Bronze générées par le pipeline.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style='whitegrid')
```

```python
restaurants = pd.read_json(
    '../data/bronze/restaurants.json',
    lines=True
)

clients = pd.read_json(
    '../data/bronze/clients.json',
    lines=True
)

commandes = pd.read_json(
    '../data/bronze/commandes.json',
    lines=True
)

paiements = pd.read_json(
    '../data/bronze/paiements.json',
    lines=True
)

avis = pd.read_json(
    '../data/bronze/avis.json',
    lines=True
)
```

## Répartition des catégories

```python
restaurants['categorie'].value_counts()
```

```python
plt.figure(figsize=(12,6))
restaurants['categorie'].value_counts().plot(kind='bar')
plt.title('Restaurants par catégorie')
plt.xlabel('Catégorie')
plt.ylabel('Nombre')
plt.tight_layout()
plt.show()
```

## Top 10 restaurants par nombre d'avis

```python
restaurants.sort_values(
    by='review_count',
    ascending=False
).head(10)
```

## Chiffre d'affaires par restaurant

```python
ca_restaurant = (
    commandes
    .groupby('restaurant_id')['montant']
    .sum()
    .reset_index()
)

ca_restaurant = ca_restaurant.merge(
    restaurants[['restaurant_id', 'name']],
    on='restaurant_id'
)

ca_restaurant.sort_values(
    'montant',
    ascending=False
).head(10)
```

## Évolution des commandes

```python
commandes['date_commande'] = pd.to_datetime(
    commandes['date_commande']
)

commandes['mois'] = commandes['date_commande'].dt.to_period('M')

monthly_orders = (
    commandes.groupby('mois')
    .size()
)

monthly_orders.plot(
    figsize=(14,6),
    marker='o'
)

plt.title('Nombre de commandes par mois')
plt.ylabel('Commandes')
plt.show()
```
